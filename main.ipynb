```python
train_data_mnist = datasets.MNIST(
    root="data",
    train = True,
    download = True,
    transform= ToTensor(),
    target_transform= None
)

test_data_mnist= datasets.MNIST(
    root= "data",
    train = False,
    download=True,
    transform=ToTensor()

)
```

```python
from torch.utils.data import ConcatDataset, Dataset

class DataWrapper(Dataset):

    def __init__(self , original_dataset,domain_id):

        self.original_dataset = original_dataset
        self.domain_id = domain_id

    def __len__(self):

        return len(self.original_dataset)

    def __getitem__(self, idx):

        img, label = self.original_dataset[idx]
        return img, label, self.domain_id
    
```

```python
mnist_train_wrapped = DataWrapper(train_data_mnist, domain_id=0)
svhn_train_wrapped = DataWrapper(svhn_data_train, domain_id=1)
fmnist_train_wrapped = DataWrapper(train_data_FMNIST, domain_id=2)
cifar_train_wrapped = DataWrapper(cifar_data_train, domain_id=3)
```

```python
mnist_test_wrapped = DataWrapper(test_data_mnist, domain_id=0)
svhn_test_wrapped = DataWrapper(svhn_data_test, domain_id=1)
fmnist_test_wrapped = DataWrapper(test_data_FMNIST, domain_id=2)
cifar_test_wrapped = DataWrapper(cifar_data_test, domain_id=3)
```

```python
unified_train_dataset = ConcatDataset([
    mnist_train_wrapped, 
    svhn_train_wrapped, 
    fmnist_train_wrapped, 
    cifar_train_wrapped
])

unified_test_dataset = ConcatDataset([
    mnist_test_wrapped, 
    svhn_test_wrapped, 
    fmnist_test_wrapped, 
    cifar_test_wrapped
])
```

```python
from torch import nn
```

```python
# Made the baseline mode to which our route router will be compared to :

class baseLineModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.block1 = nn.Sequential(
            nn.Conv2d(
                in_channels=3,
                out_channels=64,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.block2 = nn.Sequential(
            nn.Conv2d(
                in_channels=64,
                out_channels=128,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.block3 = nn.Sequential(
            nn.Conv2d(
                in_channels=128,
                out_channels=256,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.block4 = nn.Sequential(
            nn.Conv2d(
                in_channels=256,
                out_channels=512,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.Gap = nn.AdaptiveAvgPool2d((1, 1))

        self.domain_head = nn.Linear(
            in_features=512,
            out_features=4
        )

        self.class_head = nn.Linear(
            in_features=512,
            out_features=10
        )

    def forward(self, x):

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)

        x = self.Gap(x)

        x = torch.flatten(x, 1)

        domain_output = self.domain_head(x)
        class_output = self.class_head(x)

        return domain_output, class_output
```

```python
model_base = baseLineModel().to(device) 
```

```python
import os

unified_train_loader = torch.utils.data.DataLoader(
    unified_train_dataset,
    batch_size=64,
    shuffle=True,
    num_workers= os.cpu_count()
)

unified_test_loader = torch.utils.data.DataLoader(
    unified_test_dataset,   
    shuffle=False,
    batch_size=64,
    num_workers= os.cpu_count()
)
```

```python
def train_baseline(model : nn.Module,
                   loss_fn_domain : nn.Module,
                   optimizer : torch.optim.Optimizer,
                   data_loader : torch.utils.data.DataLoader,
                   device : torch.device,
                   loss_fn_class : nn.Module,
                   epochs : int = 10):

    for epoch in range(epochs):

        model.train()

        train_loss = 0.0

        for images,domain,classses in data_loader :

            images = images.to(device)
            domain = domain.to(device)
            classses = classses.to(device)

            y = model(images)

            loss_domain = loss_fn_domain(y[0],domain)

            loss_class = loss_fn_class(y[1],classses)

            total_loss = loss_domain + loss_class

            train_loss += total_loss.item()

            optimizer.zero_grad()

            total_loss.backward()

            optimizer.step()

        train_loss /= len(data_loader)

        if epoch % 5 == 0 :
                print(f"Epoch [{epoch+1}/{epochs}], Loss: {train_loss:.4f}")
```

```python
def test_baseline(model : nn.Module,
                  loss_fn_domain : nn.Module,
                  loss_fn_class : nn.Module,
                  data_loader : torch.utils.data.DataLoader,
                  device : torch.device):

    model.eval()
    with torch.inference_mode():

        test_loss = 0.0

        for images,domain,classses in data_loader :

            images = images.to(device)
            domain = domain.to(device)
            classses = classses.to(device)

            y = model(images)

            loss_domain = loss_fn_domain(y[0],domain)

            loss_class = loss_fn_class(y[1],classses)

            total_loss = loss_domain + loss_class

            test_loss += total_loss.item()

        test_loss /= len(data_loader)

        print(f"Test Loss: {test_loss:.4f}")
        
```

```python
loss_fn_domain = nn.CrossEntropyLoss(label_smoothing=0.1)
loss_fn_class = nn.CrossEntropyLoss(label_smoothing=0.1)

optimizer = torch.optim.AdamW(model_base.parameters(), lr=3e-4)
```

```python
# Made the experts model which will be used the by the model which are connected by routes

class experts(nn.Module):

    def __init__(self):
        super().__init__()

        self.block1 = nn.Sequential(
            nn.Conv2d(
                in_channels=3,
                out_channels=32,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.block2 = nn.Sequential(
            nn.Conv2d(
                in_channels=32,
                out_channels=64,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.block3 = nn.Sequential(
            nn.Conv2d(
                in_channels=64,
                out_channels=128,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.block4 = nn.Sequential(
            nn.Conv2d(
                in_channels=128,
                out_channels=256,
                stride=1,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.Gap = nn.AdaptiveAvgPool2d((1, 1))

        self.class_head = nn.Linear(
            in_features=256,
            out_features=10
        )

    def forward(self, x):

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)

        x = self.Gap(x)

        x = torch.flatten(x, 1)
        class_output = self.class_head(x)

        return  class_output
```
